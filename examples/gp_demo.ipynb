# Free local GP scheduling example
This executed notebook demonstrates the tested prototype. Durations/noise are synthetic; no cloud or paid service is called.

```python
from pathlib import Path
from qcgp import HybridCompiler, GPSchedulingPass
from qcgp.core import benchmark, topology
from qiskit.transpiler import PassManager
logical = benchmark('qaoa', 3, 101)
coupling = topology(3, 'line')
result = HybridCompiler(coupling).compile(logical)
print('Selected candidate:', result.selected_candidate)
print('Guarantee:', result.guarantee)
```

```python
manager = PassManager([GPSchedulingPass(coupling, solver='ECOS')])
output = manager.run(result.circuit)
schedule = manager.property_set['gp_schedule']
print('Makespan:', round(schedule.makespan, 6))
print('Free solver:', schedule.solver_used)
print('Status:', schedule.status)
```

## Interpretation
The scheduling optimum also follows from a weighted critical path. GP does not improve that fixed-order optimum. The compiler searches a finite candidate set; its guarantee is limited to that set. See the saved evidence for the complete benchmark.
